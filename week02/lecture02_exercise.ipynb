**Task 3 (stretch):** Build a **grouped bar chart** comparing 2 sub-factors (e.g. `GDP_per_capita` and `Freedom`) across the 5 most populated regions. Use colour meaningfully and write an insight title.

Regions to include: `'Western Europe'`, `'Latin America'`, `'East Asia'`, `'Sub-Saharan Africa'`, `'South Asia'`


```python
# Stretch goal — grouped bar chart
regions = ['Western Europe', 'Latin America and Caribbean', 'East Asia',
           'Sub-Saharan Africa', 'South Asia']
factors = ['Social_Support', 'Freedom']
region_factors = (df[df['Region'].isin(regions)]
                  .groupby('Region')[factors]
                  .mean()
                  .reset_index())

region_factors['Gap'] = region_factors['Social_Support'] - region_factors['Freedom']
region_factors = region_factors.sort_values('Social_Support', ascending=False)
print(region_factors.round(3)) 

```

```python
long_df = region_factors.melt(id_vars='Region', value_vars=factors,
                              var_name='Factor', value_name='Score')

long_df['Factor'] = long_df['Factor'].replace({'Social_Support': 'Social support',
                                               'Freedom': 'Freedom'})

biggest = region_factors.loc[region_factors['Gap'].idxmax()]
title = (f"In {biggest['Region']}, social support beats freedom by {biggest['Gap']:.2f} points "
         f"— the widest gap of the five regions")
fig = px.bar(
    long_df,
    x='Score', y='Region',                 
    color='Factor',                        
    barmode='group',                       
    orientation='h',
    text='Score',                          
    color_discrete_map={'Social support': '#2E75B6',   
                        'Freedom': '#ED7D31'},         
    category_orders={'Region': region_factors['Region'].tolist()}, 
    labels={'Score': 'Average score (0–1 scale)', 'Region': ''},
    title=title,
    template='plotly_white',               
    height=500,
)
fig.update_traces(texttemplate='%{text:.2f}', textposition='outside')

fig.update_layout(
    font=dict(size=13),
    xaxis=dict(range=[0, 1.05],            
               showgrid=False),
    yaxis=dict(autorange='reversed',       
               showgrid=False),
    legend=dict(title='', orientation='h', 
                yanchor='bottom', y=1.02, xanchor='left', x=0),
    margin=dict(r=40),
)

fig.show()
```
